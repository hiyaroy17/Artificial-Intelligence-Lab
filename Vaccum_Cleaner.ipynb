```python

A = "Dirty"
B = "Dirty"

location = "A"

while A == "Dirty" or B == "Dirty":

    if location == "A":
        if A == "Dirty":
            print("Cleaning Room A")
            A = "Clean"
        else:
            print("Move to Room B")
            location = "B"

    else:
        if B == "Dirty":
            print("Cleaning Room B")
            B = "Clean"
        else:
            print("Move to Room A")
            location = "A"

print("Both rooms are CLEAN!")
```
