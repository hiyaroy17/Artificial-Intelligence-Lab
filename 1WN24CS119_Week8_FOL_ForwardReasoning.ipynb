```python
# Initial Facts
facts = {
    "American(John)",
    "Enemy(B,America)",
    "Missile(M1)",
    "Owns(B,M1)"
}

changed = True
while changed:
    changed = False

    # Rule 1: Enemy(x,America) -> Hostile(x)
    if "Enemy(B,America)" in facts and "Hostile(B)" not in facts:
        facts.add("Hostile(B)")
        changed = True

    # Rule 2: Missile(x) -> Weapon(x)
    if "Missile(M1)" in facts and "Weapon(M1)" not in facts:
        facts.add("Weapon(M1)")
        changed = True

    # Rule 3: Missile(x) and Owns(B,x) -> Sells(John,x,B)
    if "Missile(M1)" in facts and "Owns(B,M1)" in facts:
        if "Sells(John,M1,B)" not in facts:
            facts.add("Sells(John,M1,B)")
            changed = True

    # Rule 4:
    # American(p) and Weapon(q) and Sells(p,q,r) and Hostile(r) -> Criminal(p)
    if ("American(John)" in facts and
        "Weapon(M1)" in facts and
        "Sells(John,M1,B)" in facts and
        "Hostile(B)" in facts):
        if "Criminal(John)" not in facts:
            facts.add("Criminal(John)")
            changed = True

print("Derived Facts:")
for fact in facts:
    print(fact)

if "Criminal(John)" in facts:
    print("\nQuery Proved: Criminal(John)")
else:
    print("\nQuery Not Proved")
```
