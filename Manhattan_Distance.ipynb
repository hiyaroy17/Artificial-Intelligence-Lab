```python

import heapq

def create_goal():
    return (
        (3, 2, 1),
        (6, 5, 4),
        (8, 7, 0)
    )

def manhattan_distance(state, goal):
    distance = 0

    for i in range(3):
        for j in range(3):

            if state[i][j] != 0:

                for x in range(3):
                    for y in range(3):

                        if goal[x][y] == state[i][j]:
                            distance += abs(i - x) + abs(j - y)

    return distance


def get_neighbors(state):
    neighbors = []

    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                x, y = i, j

    moves = [(1, 0), (-1, 0), (0, 1), (0, -1)]

    for dx, dy in moves:
        nx = x + dx
        ny = y + dy

        if 0 <= nx < 3 and 0 <= ny < 3:

            new_state = [list(row) for row in state]

            new_state[x][y], new_state[nx][ny] = \
                new_state[nx][ny], new_state[x][y]

            neighbors.append(tuple(tuple(row) for row in new_state))

    return neighbors


def a_star(start, goal):

    # (f, g, state, path)
    pq = []

    h = manhattan_distance(start, goal)

    heapq.heappush(pq, (h, 0, start, [start]))

    visited = set()

    while pq:

        f, g, state, path = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        if state == goal:
            return path

        for next_state in get_neighbors(state):

            if next_state not in visited:

                new_g = g + 1
                new_h = manhattan_distance(next_state, goal)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (new_f, new_g, next_state, path + [next_state])
                )

    return None


# Initial State
start = (
    (3, 2, 1),
    (6, 0, 4),
    (8, 5, 7)
)

goal = create_goal()

print("Initial State:")
for row in start:
    print(row)

print("\nGoal State:")
for row in goal:
    print(row)


path = a_star(start, goal)

print("\nA* Solution:")

for step, state in enumerate(path):
    print("\nStep", step)

    for row in state:
        print(row)

print("\nTotal Moves =", len(path) - 1)
```
