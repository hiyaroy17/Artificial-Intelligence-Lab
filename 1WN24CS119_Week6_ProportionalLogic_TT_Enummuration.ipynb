```python
import itertools

def evaluate(expr, model):
    return eval(expr, {}, model)

def entails(kb, query, symbols):
    for values in itertools.product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))

        if all(evaluate(s, model) for s in kb):
            if not evaluate(query, model):
                return False
    return True


kb = [
    "P == Q",
    "Q or R",
    "(not P) <= R"
]

symbols = ["P", "Q", "R"]

print("P\tQ\tR\tP<->Q\tQ OR R\t¬P->R\tKB")
print("-" * 55)

for values in itertools.product([True, False], repeat=3):
    model = dict(zip(symbols, values))

    results = [evaluate(s, model) for s in kb]
    kb_true = all(results)

    print(
        model["P"], model["Q"], model["R"],
        *results, kb_true, sep="\t"
    )

queries = {
    "Q": "Q",
    "P -> R": "(not P) or R",
    "R -> Q": "(not R) or Q"
}

print("\nEntailment Results:")
for name, query in queries.items():
    print(f"Does KB entail {name}? {entails(kb, query, symbols)}")
```
